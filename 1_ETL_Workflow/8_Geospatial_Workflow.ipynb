# IDSW v2.0.0 Workflow: Geospatial Analysis & Mapping (`idsw.geospatial`)

This notebook demonstrates the **IDSW 2.0.0** `geospatial` module:
- Geospatial data loading & CRS transformation (`load_geospatial_data`, `convert_crs`, `export_geospatial_data`).
- Spatial operations (`spatial_join`, `create_spatial_buffers`, `dissolve_geometries`, `calculate_spatial_features`).
- Visualization (`plot_choropleth` with `plotly=True`, `create_interactive_map`, `create_marker_cluster_map`, `create_spatial_heatmap`).

```python
import pandas as pd
from idsw import *
from idsw.geospatial import *

ControlVars.show_plots = True
ControlVars.show_results = True
```

```python
# Example 1: Load GeoJSON or CSV with latitude/longitude
sites_df = pd.DataFrame({
    'site_name': ['Plant A', 'Plant B', 'Plant C'],
    'lat': [-23.5505, -22.9068, -19.9167],
    'lon': [-46.6333, -43.1729, -43.9345],
    'emissions_tons': [120.5, 85.2, 145.8]
})

gdf_sites = load_geospatial_data(df=sites_df, lat_col='lat', lon_col='lon', crs='EPSG:4326')

# Convert CRS to projected metric system (SIRGAS 2000 / UTM Zone 23S) and compute buffers
gdf_utm = convert_crs(gdf_sites, target_crs='EPSG:31983')
gdf_buffers = create_spatial_buffers(gdf_utm, distance=5000) # 5 km buffer
gdf_features = calculate_spatial_features(gdf_buffers)
```

```python
# Example 2: Interactive Plotly Choropleth & Folium Web Map
fig_choro, _ = plot_choropleth(
    gdf_buffers,
    column='emissions_tons',
    title='Industrial Sites 5km Buffer Zones - Emissions',
    plotly=True,
    export_png=True,
    file_name='industrial_choropleth'
)

folium_map = create_interactive_map(
    gdf_sites,
    popup_cols=['site_name', 'emissions_tons'],
    export_html=True,
    file_name='industrial_sites_map.html'
)
```
