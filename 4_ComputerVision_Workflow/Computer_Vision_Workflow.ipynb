```python
# 1. Synthetic Image Processing Example
img = np.zeros((128, 128), dtype=np.uint8)
img[32:96, 32:96] = 200
img = img + np.random.randint(0, 30, size=(128, 128), dtype=np.uint8)

denoised = denoise_image(img, method='gaussian', sigma=1.0)
thresh_mask = apply_thresholding(denoised, method='otsu')
edges = detect_edges(denoised, method='sobel')
opened = apply_morphological_operation(thresh_mask, operation='opening', kernel_size=3)
```

```python
# 2. Biomedical 3D Volume Analysis Example
ct_volume = np.random.normal(-500, 100, size=(16, 64, 64))
ct_volume[4:12, 20:44, 20:44] = 50  # Soft tissue structure

windowed = apply_hounsfield_window(ct_volume, window_center=40, window_width=400)
labeled_vol, n_objs = segment_3d_volume(windowed, threshold=0.4)
morph_df = measure_morphological_features(labeled_vol, intensity_volume=ct_volume)
```
