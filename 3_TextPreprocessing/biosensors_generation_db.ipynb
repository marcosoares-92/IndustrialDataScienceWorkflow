```python
import os
# Directory containing biosensor PDF documents
directory = '/content/drive/MyDrive/biosensors'

if os.path.exists(directory):
    files = os.listdir(directory)
    file_paths = [os.path.join(directory, f) for f in files if f.lower().endswith('.pdf')]
    print(f'Found {len(file_paths)} PDF documents.')
else:
    print(f'Directory {directory} not found. Using sample file paths.')
    file_paths = []
```

## 1. Automated Text Extraction with IDSW
	ext_extraction loads all documents, extracts metadata, concatenates text with standard delimiters, and creates the text database.

```python
from idsw.datafetch.texts import (
    text_extraction,
    create_txt_db,
    read_txt_database,
    convert_text_db_format
)

if len(file_paths) > 0:
    # Extract texts and metadata, exporting as TXT
    texts, metadata, file_names = text_extraction(
        file_paths = file_paths,
        doc_separator = '\\n\\n-----\\n\\n',
        output_text_path = 'biosensors_db.txt',
        output_meta_path = 'biosensors_metadata_db.txt',
        export_as_pdf = False
    )
    print(f'Total documents extracted: {len(texts)}')
    print('Sample extracted text preview:', texts[0][:300] if len(texts) > 0 else 'N/A')
```

## 2. Exporting Database as PDF with export_as_pdf=True`nUsing the optimized create_txt_db function, we can export the database directly to PDF format.

```python
if len(file_paths) > 0:
    # Export directly to PDF database
    create_txt_db(
        file_path = 'biosensors_db.pdf',
        strings_to_save = texts,
        doc_separator = '\\n\\n-----\\n\\n',
        export_as_pdf = True
    )
    print('Exported biosensors_db.pdf successfully.')
```

## 3. Format Conversion: TXT <-> PDF with convert_text_db_format`nEasily convert existing databases between .txt and .pdf formats.

```python
# Convert existing TXT database to PDF
if os.path.exists('biosensors_db.txt'):
    convert_text_db_format(
        file_path = 'biosensors_db.txt',
        output_format = 'pdf',
        exported_file_path = 'biosensors_converted_from_txt.pdf'
    )

# Convert PDF database back to TXT
if os.path.exists('biosensors_db.pdf'):
    convert_text_db_format(
        file_path = 'biosensors_db.pdf',
        output_format = 'txt',
        exported_file_path = 'biosensors_converted_from_pdf.txt'
    )
```

## 4. Reading Database into Python with ead_txt_database`nSeamlessly read back databases saved in either TXT or PDF format.

```python
if os.path.exists('biosensors_db.txt'):
    read_docs = read_txt_database('biosensors_db.txt', split_strings = True)
    print(f'Read {len(read_docs)} documents from biosensors_db.txt.')

if os.path.exists('biosensors_db.pdf'):
    read_pdf_docs = read_txt_database('biosensors_db.pdf', split_strings = True)
    print(f'Read {len(read_pdf_docs)} documents from biosensors_db.pdf.')
```
